### Step 1: a dict is one labeled item

```python
message = {'role': 'user', 'content': 'Hello'}

print(message)
print(message['role'])
print(message['content'])
```

### Step 2: a list holds many items in order

```python
message = []
print(message)

message.append({'role': 'user', 'content': 'Hello'})
print(message)

message.append({'role': 'assistant', 'content': 'Hi there!'})
print(message)
print(len(message))
```

### Step 3: reading items from a list

```python
messages = [
    {"role": "user", "content": "Hi"},
    {"role": "assistant", "content": "Hello!"},
    {"role": "user", "content": "My name is Suraj"},
    {"role": "assistant", "content": "Nice to meet you!"},
]

print(messages[0])
print(messages[-1])
print(messages[-2:])
```

### Step 4: a loop reads every item

```python
for m in messages:
    print(f"{m['role']} ---> {m['content']}")
```

### Step 5: a fake model that only sees what you send

```python
def fake_model(msgs):
    return f"I can see {len(msgs)} messages"

notebook = []

notebook.append({'role': 'user', 'content': 'Hi' })
reply = fake_model(notebook)
print(reply)
notebook.append({'role': 'assistant', 'content': reply})

notebook.append({"role": "user", "content": "Hello again"})
reply = fake_model(notebook)
print(reply)
```

```python
# 1st print - I can see 1 messages
# 2nd print - I can see 3 messages
```

### Step 6: the same thing in a loop

```python
notebook = []

while True:
    user_text = input('You: ')
    if user_text.lower() == 'exit':
        break
    notebook.append({'role': 'user', 'content': user_text})
    reply = fake_model(notebook)
    print('Bot: ', reply)
    notebook.append({'role': 'assistant', 'content': reply})
```

### Step 7: trimming

```python
notebook = []

while True:
    user_text = input('You: ')
    if user_text.lower() == 'exit':
        break
    notebook.append({'role': 'user', 'content': user_text})
    recent = notebook[-4:]
    reply = fake_model(recent)
    print('Bot: ', reply)
    notebook.append({'role': 'assistant', 'content': reply})
```
